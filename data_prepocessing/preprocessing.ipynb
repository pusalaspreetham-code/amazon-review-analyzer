Import and loading of dataset

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../amazon_reviews.csv")

print("Dataset shape:", df.shape)
print("Unique products:", df["product_id"].nunique())
print("Unique review IDs:", df["review_id"].nunique())

display(df.head())
```

Check missing values and placeholders

```python
print("Missing values per column:")
display(df.isnull().sum().sort_values(ascending=False))

print("\n Placeholder counts")
for col in df.columns:
    count=df[col].astype("string").str.strip().str.lower().eq("not available").sum()
    if count>0:
        print(f"{col}:{count}")
```

Investigate duplicate review IDs

```python
duplicate_ids = df[df["review_id"].duplicated(keep=False)]

display(duplicate_ids.sort_values("review_id").head(20))
```

Analyze duplicate review IDs

```python
# How many unique review texts and ratings exist per review ID?
duplicate_analysis = (
    df.groupby("review_id")
      .agg(
          occurrences=("review_id", "size"),
          unique_products=("product_id", "nunique"),
          unique_texts=("review_text", "nunique"),
          unique_ratings=("review_rating", "nunique"),
          unique_prices=("price", "nunique")
      )
      .query("occurrences > 1")
      .sort_values("occurrences", ascending=False)
)

display(duplicate_analysis.head(20))
```

Remove duplicate review IDs

```python
df = df.drop_duplicates(
    subset=["review_id"],
    keep="first"
).reset_index(drop=True)
print("Rows after:", len(df))
```

Check missing values

```python
missing_values = df.isnull().sum()

missing_values = missing_values[missing_values > 0].sort_values(
    ascending=False
)

display(missing_values)
```

check missing review title and text

```python
print("Missing review text:", df["review_text"].isna().sum())
print("Missing review title:", df["review_title"].isna().sum())

print(
    "Blank review text:",
    df["review_text"].fillna("").str.strip().eq("").sum()
)

print(
    "Blank review title:",
    df["review_title"].fillna("").str.strip().eq("").sum()
)
```

delete missing values

```python
df = df.dropna(subset=["review_text", "review_title"])

df = df[
    df["review_text"].str.strip().ne("") &
    df["review_title"].str.strip().ne("")
].reset_index(drop=True)

print("Dataset shape after removing rows:", df.shape)
```

Replace placeholders with proper missing values

```python
df=df.replace(r"(?i)^\s*not available\s*$",np.nan,regex=True)

missing_values = df.isnull().sum()
display(missing_values[missing_values > 0].sort_values(ascending=False))
```
