Amazon Reviews - Data Preprocessing

1. Import libraries and load dataset

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../amazon_reviews.csv")

print("Dataset shape:", df.shape)
df.head()
```

2. Check the dataset

```python
print(df.info())
print(df.isnull().sum())
```

3. Check Duplicate values

```python
# Check duplicate reviews
duplicate_count = df.duplicated(
    subset=["review_id", "product_id"]
).sum()

print("Number of Duplicate Reviews:", duplicate_count)
```

3. Convert Numeric Columns

```python
# Convert columns into numeric format
numeric_columns = [
    "price", "original_price", "product_rating",
    "total_product_reviews", "five_star_review_count",
    "four_star_review_count", "three_star_review_count",
    "two_star_review_count", "one_star_review_count",
    "review_rating", "helpful_votes", "discount_percentage"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

4. Remove Missing and Invalid Values

```python
# Remove rows containing missing values
df = df.dropna()

# Keep valid prices and ratings
df = df[
    (df["price"] > 0) &
    (df["original_price"] >= df["price"]) &
    (df["product_rating"].between(1, 5)) &
    (df["review_rating"].between(1, 5)) &
    (df["total_product_reviews"] > 0) &
    (df["helpful_votes"] >= 0)
]

print("Dataset Shape After Cleaning:", df.shape)
```
